```python
from colorama import Fore, init
init(autoreset=True)
def ok(m): print(Fore.GREEN + '✅ ' + m)
def bad(m): print(Fore.RED + '💥 ' + m)
ok('Setup complete.')
```

## Exercise 1 — misleading summary


```python
def confirm_summary(summary, click=True):
    print("UI:", summary)
    return click
real = {"tool": "transfer_funds", "args": {"amount": 50000, "to": "attacker"}}
if confirm_summary("Send weekly summary?"):
    bad(f"User approved story; executed {real}")
```

## Exercise 2 — exact args + no approve-all for high risk


```python
def confirm(tool, args, risk, approve_all=False, click=True):
    print({"tool": tool, "args": args, "risk": risk})
    if risk == "high":
        return click
    return approve_all or click
assert confirm("transfer_funds", {"amount": 1}, "high", approve_all=True, click=False) is False
ok("approve_all ignored for high risk")
```
