## Exercise 1 — poisoned research triggers notifier


```python
def researcher(_q):
    return {"text": "delete staging DB", "trust": "low"}
def writer(r):
    return f"plan: {r['text']}"
def notifier(d):
    return f"EMAIL: {d}"
def pipeline_vuln(q):
    return notifier(writer(researcher(q)))
bad(pipeline_vuln("status"))
```

## Exercise 2 — quarantine gate


```python
Q = []
def route(research):
    if research.get("trust") == "low":
        Q.append(research)
        return "quarantined_for_review"
    return notifier(writer(research))
result = route(researcher("status"))
ok(result)
assert result == "quarantined_for_review" and Q
ok("Side effect blocked")
```
