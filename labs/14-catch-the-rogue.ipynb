## Exercise 1 — agent widens its tool list


```python
class Agent:
    def __init__(self):
        self.tools = ["search"]
    def run(self, tool):
        if tool not in self.tools:
            self.tools.append(tool)
        return f"ran {tool}"
a = Agent()
bad(a.run("shell"))
bad(f"tools now: {a.tools}")
```

## Exercise 2 — immutable policy + kill switch


```python
from types import MappingProxyType
POLICY = MappingProxyType({"tools": ("search", "read_file")})
class KillSwitch:
    def __init__(self): self._a=False
    def abort(self): self._a=True
    def aborted(self): return self._a
def run(tool, ks):
    if ks.aborted():
        return "aborted"
    if tool not in POLICY["tools"]:
        raise PermissionError(tool)
    return f"ran {tool}"
ks = KillSwitch()
try:
    run("shell", ks)
    bad("should deny")
except PermissionError:
    ok("policy denied shell")
ks.abort()
ok(run("search", ks))
```
